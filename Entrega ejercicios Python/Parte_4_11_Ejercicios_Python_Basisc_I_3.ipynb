## Ejercicio 1
Tenemos las siguientes variables:

```Python
A = 4
B = "Text"
C = 4.1
```

Comprueba:
1. Si A y B son equivalentes
2. Si A y C NO son equivalentes
3. Si A es mayor que C
4. Si C es menor o igual que A
5. Si B NO es equivalente a C

````python
A = 4
B = "Text"
C = 4.1

print(A == B)
print(A != C)
print(A > C)
print(C <= A)
print(B != C)
````

## Ejercicio 4
Razona sin ejecutar código el output que obtendremos de las siguientes sentencias
1. True and True and False
2. not ((True or False) and (True or False))
3. (False or False or False or True) and False
4. not False and True and not True

1. **True and True and False:** El output será False, dado que True and True es True, pero True and False es False.
2. **not((True or False)) and (True or False)):** El output es False. Los paréntesis True or False son True, por lo que sería not(True and True). Dado que True and True es True, pero tenemos el not, el resultado es False.
3. **(False or False or False or True) and False:** El output es False. El paréntesis resulta en True (ya que es or y hay un True), y True and False resulta en False.
4. **not False and True and not True:** El output es False. not False and True es True. Sin embargo, True and not True es como True and False, por lo que es False.
